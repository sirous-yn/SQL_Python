```python
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'], errors='coerce')
df['job_posted_month'] = df['job_posted_date'].dt.month_name()
```

```python
#month_num = df['job_posted_date'].dt.month
months = df['job_posted_month'].unique()
# dictionary of data frames,, making new data frames to practice pd.concat
dict_month = {month:df[df['job_posted_month'] == month] for month in months}
dict_month['January']
```

```python
df_q1 = pd.concat([dict_month['January'], dict_month['February'], dict_month['March']], ignore_index=True)
df_q1['job_posted_month'].value_counts().plot(kind='barh')
```

To CSV

```python
from pathlib import Path

file = Path('quarter1.csv')

if not file.exists():
    df_q1.head(5).to_csv(file, index=False)
# index=False prevents pandas from saving DataFrame's index as an additional column
```
