```python
import pandas as pd
import numpy as np
from datasets import load_dataset
import matplotlib.pyplot as plt

# loading data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
```

Index Management

```python
df.sample(2)
# indx has no name
```

```python
df.index
```

```python
# giving indx column a name
df.index.name = 'job_index'
#df.sort_values('job_index', ascending=False) this method does not work, though we gave the index column a name, still can not 
# sort it using sortvalues and must use sort_index for this column
df.sort_index(ascending=False)
df.sample(2)
```

Index Reset

```python
df_usa = df.loc[df['job_country']=='United States',['job_title_short','job_country']]
#indices are missing some values as a filter is applied
df_usa
```
